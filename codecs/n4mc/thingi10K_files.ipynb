```python
import os
import itertools

save_dir = "/mnt/datadrive/ChromeDownloads/Thingi10K/meshes"
os.makedirs(save_dir, exist_ok=True)

# Iterate through dataset with constraints
for i, entry in enumerate(itertools.islice(thingi10k.dataset(num_vertices=(20000, 30000), closed=True), 100)):
    
    vertices, facets = thingi10k.load_file(entry["file_path"])

    # Convert to Open3D TriangleMesh
    mesh = o3d.geometry.TriangleMesh()
    mesh.vertices = o3d.utility.Vector3dVector(vertices)
    mesh.triangles = o3d.utility.Vector3iVector(facets)

    # Optional: compute normals for better visualization
    mesh.compute_vertex_normals()

    # Save as .obj file
    obj_path = os.path.join(save_dir, f"mesh_{i:03d}.obj")
    o3d.io.write_triangle_mesh(obj_path, mesh)

    print(f"Saved: {obj_path}  ({len(vertices)} vertices, {len(facets)} faces)")
```
