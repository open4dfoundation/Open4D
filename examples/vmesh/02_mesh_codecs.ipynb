# Mesh USDC → .vmesh → mesh USDC

Set your input and choose `tvmc`, `tsmc`, `vdmc`, `faster_vdmc` or `n4mc`.
Configure its [native backend](../../README.md) first. These examples encode geometry only.

```python
from pathlib import Path
import open4d
from open4d.codec import inspect_vmesh

source = Path("input.usdc")
output = Path("output/mesh")
codec = "tvmc"
encode_options = {}
decode_options = {}
```

For N4MC on CPU, set `codec="n4mc"` and both option dictionaries to `{"device": "cpu"}`.
Add `resolution` and `epochs` to `encode_options` to control training.

```python
encoded = open4d.encode(
    source, output / "motion.vmesh", codec=codec, overwrite=True, **encode_options)
print(inspect_vmesh(encoded)["frame_count"], "frames encoded")
```

```python
with open4d.load(encoded, options=decode_options) as decoded:
    open4d.save(decoded, output / "reconstructed.usdc", overwrite=True)
    print(len(decoded), "frames decoded")
```

Existing O4D `.tvmc` / `.tsmc` directories or `.n4d` files can be migrated without re-encoding.
Run the next cell only when you have such an artifact.

```python
from open4d.codec import pack_vmesh

pack_vmesh("existing.tvmc", output / "migrated.vmesh", overwrite=True)
```
