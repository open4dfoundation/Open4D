# Native Gaussian runs → .vmesh

Run the section for your codec after setting its paths. Use the configuration that trained
the run. Packing needs no CUDA; the final decoding cell requires the codec's configured runtime.

```python
from pathlib import Path
import open4d

output = Path("output/gaussians")
```

Vega

```python
with open4d.import_native("runs/capture.vega", codec="vega") as native:
    open4d.save(native, output / "vega.vmesh", overwrite=True)
    print(len(native), "Vega frames packed")
```

QUEEN

```python
with open4d.import_native("runs/queen", codec="queen", config="runs/queen/config.yaml") as native:
    open4d.save(native, output / "queen.vmesh", overwrite=True)
    print(len(native), "QUEEN frames packed")
```

3DGStream. The run must contain its initial model, NTC checkpoints and architecture configuration.

```python
with open4d.import_native("runs/3dgstream", codec="3dgstream",
                         ntc_config="runs/3dgstream/ntc_config.json") as native:
    open4d.save(native, output / "3dgstream.vmesh", overwrite=True)
    print(len(native), "3DGStream frames packed")
```

Decode one saved artifact. Set `OPEN4D_<CODEC>_ROOT` and `OPEN4D_<CODEC>_PYTHON` for its native CUDA runtime.

```python
with open4d.load(output / "queen.vmesh") as native:
    frames = native.decode()
    print(len(frames), "frames;", len(frames[0].positions), "Gaussians in the first frame")
```
