# Native ReRF run → .vmesh

Set the compressed run path. It must contain its entropy files, motion, shared RGB network
and resolved `decoder_config.json`, including rendering bounds. Packing needs no CUDA.

```python
import json
from pathlib import Path
import open4d
from open4d.codec import inspect_vmesh

source = Path("runs/rerf")
output = Path("output/rerf")
```

```python
with open4d.import_native(source, codec="rerf", config=source / "decoder_config.json") as native:
    artifact = open4d.save(native, output / "rerf.vmesh", overwrite=True)
print(inspect_vmesh(artifact)["frame_count"], "frames packed")
```

Decode using `OPEN4D_RERF_ROOT` and `OPEN4D_RERF_PYTHON` for the native CUDA runtime.

```python
with open4d.load(artifact) as native:
    fields = native.decode()
print(fields[0].density.shape, fields[0].features.shape)
```

Render a calibrated camera. `camera.json` contains `camera_id`, `width`, `height`,
`fx`, `fy`, `cx`, `cy` and a 4×4 `c2w` matrix for this run.

```python
import matplotlib.pyplot as plt

camera = json.loads(Path("camera.json").read_text())
with open4d.load(artifact) as native:
    images = native.render(camera)
plt.imshow(images[0])
plt.axis("off");
```
