# Mesh USDC → .o4d → mesh USDC

Set your input and choose `klt`, `n4mc`, `qndf`, `qndf-int8`, `tvmc`, `tsmc`, `vdmc` or `faster_vdmc`.
Configure its [research backend](../../README.md) first. These examples encode geometry only.

```python
from pathlib import Path
import open4d
from open4d.codec import inspect_o4d

source = Path("input.usdc")
output = Path("output/mesh")
codec = "tvmc"
encode_options = {}
decode_options = {}
```

For N4MC on CPU, set `codec="n4mc"` and both option dictionaries to `{"device": "cpu"}`.
Add `resolution` and `epochs` to `encode_options` to control training.

```python
encoded = open4d.encode(
    source, output / "motion.o4d", codec=codec, overwrite=True, **encode_options)
print(inspect_o4d(encoded)["frame_count"], "frames encoded")
```

```python
with open4d.load(encoded, options=decode_options) as decoded:
    open4d.save(decoded, output / "reconstructed.usdc", overwrite=True)
    print(len(decoded), "frames decoded")
```

Older compressed artifacts require explicit migration. Run the next cell only when you have an older artifact; ordinary encode/decode use `.o4d`.

```python
open4d.migrate_legacy("older_artifact", output / "migrated.o4d", overwrite=True)
```
